## Why we need to Hide Personal Info:   

   Our words reveal a lot: names, addresses, habits. Hiding this info  
  * protects us from:
  Identity theft: Hackers can use personal details for financial crimes.
  * Discrimination: Biases based on online data can lead to unfair treatment.
  * Stalking and harassment: Sharing too much can attract unwanted attention.
  * Targeted advertising: Companies can exploit personal info for manipulative ads.


## Strategies of Text Anonymization
  Text anonymization involves masking or changing personal information in textual data to prevent the identification of individuals. Several strategies of anonymization can be used, including suppression, pseudonymization, noising, and generalization. Each strategy has its advantages and disadvantages, and the choice of strategy depends on the specific use case. You can read more about these strategies [here](https://www.tonic.ai/blog/anonymization-techniques-defined-transforming-real-data-into-realistic-test-data).


## Text Anonymization
We will try the Name Entity Recognition (NER) method to anonymize textual data in English.

```python
from transformers import pipeline
from transformers import AutoModelForTokenClassification, AutoTokenizer
```

```python
MODEL = "xlm-roberta-large-finetuned-conll03-english"
```

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL)
model = AutoModelForTokenClassification.from_pretrained(MODEL)
classifier = pipeline("ner", model=model, tokenizer=tokenizer)
```

**also we need to hide numbers as well**

```python
result=[]
text= "Alya told Jasmine that Andrew could pay with cash.."
entities=classifier(text)
for ent in entities:
  if ent["entity"] in ["I-PER","B-PER", "I-LOC","B-LOC"]:
      result.append(ent)

print({"text": text,"\npii": result})
```
